```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'ID':[101, 102, 103, 104],
                    'NAME':['Rahul', 'Shvani', 'Shradhha', 'Nilesh'],
                    'AGE':[34, 56,  55, 29]})
df1
```

```python
df2 = pd.DataFrame({'ID':[104, 105, 106],
                    'NAME':['Rohit', 'Pradnya', 'Yash'],
                    'AGE':[39, 22, 49]})
df2
```

```python
pd.concat([df1, df2])
```

```python
df3 = pd.DataFrame({'ID':[105, 106, 107],
                    'NAME':['Rahul', 'Shvani', 'Shradhha'],
                    'SAL':[34, 56,  55]})
df3
```

```python
pd.concat([df1, df3])
```

```python
df4 = pd.DataFrame({'ADDRESS':['Pune', 'Mumbai', 'Chennai', 'Kashmir'],
                    'DEPT':['Dev', 'DA', 'DS', 'AI/ML'],
                    'SALLARY':[45000, 56000,  48000, 35000]})
df4
```

```python
emp = pd.DataFrame({'ID':[101, 102, 103, 104],
                    'NAME':['Rahul', 'Shvani', 'Shradhha', 'Nilesh'],
                    'AGE':[34, 56,  55, 29],
                    'DID':[1, 3, 2, 5]
                    })
emp
```

```python
dept = pd.DataFrame({'DID':[1, 2, 3, 4],
                     'DNAME':['IT', 'Admin', 'HR', 'Sales']})
dept
```

```python
pd.merge(emp, dept)
```

```python
Dept2 = pd.DataFrame({'ID':[1, 2, 3, 4],
                     'DNAME':['IT', 'Admin', 'HR', 'Sales']})
Dept2
```

```python
pd.merge(emp, Dept2, left_on='ID', right_on='ID')
```

```python
pd.merge(emp, dept, how='left')
```

```python
pd.merge(emp, dept, how='right')
```

```python
pd.merge(emp, dept, how='outer')
```
